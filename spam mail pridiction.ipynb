# feature actraction

```python
# transform the test data to feature vector that can used as input to te logistic regration
feature_extration=TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)
x_train_feature=feature_extration.fit_transform(x_train)
x_test_feature=feature_extration.transform(x_test)

# convert y_train and y test values  as integir
y_train=y_train.astype('int')
y_test=y_test.astype('int')
```

```python
print(x_train_feature)
```

```python
# trainging meachine learing model
model=LogisticRegression()
```

```python
model.fit(x_train_feature,y_train)
```

```python
# evoluting the trainend model
#1 pridiction training dtaa
x_train_pridiction=model.predict(x_train_feature)
trining_dtaa_accuracy=accuracy_score(x_train_pridiction,y_train)
print('accuarcy on training data: ', trining_dtaa_accuracy)

```

```python
# pridiction on test data
x_test_pridiction=model.predict(x_test_feature)
testing_data_accuracy=accuracy_score(y_test,x_test_pridiction)
print('aaccuracy on test data: ',testing_data_accuracy)
```
