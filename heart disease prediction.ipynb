```python
x=df.drop('target',axis=1)
y=df['target']
```

```python
print(x,y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2,stratify=y) 
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
# model traing
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
# modelevolution
```

```python
# accuracy scor eon trainging data
x_train_prediction=model.predict(x_train)
training_data_accuracy=accuracy_score(x_train_prediction,y_train)
```

```python
print('accracyscoreon train data:',training_data_accuracy)
```

```python
x_test_prediction=model.predict(x_test)
testing_data_accuarcy=accuracy_score(x_test_prediction,y_test)
```

```python
print('acctracy score on test data:',testing_data_accuarcy)
```

```python
# prisision
from sklearn.metrics import precision_score
# prission for training data pridiction
peission_train=precision_score(y_train,x_train_prediction)
print('training data prission',peission_train)
prission_test=precision_score(y_test,x_test_prediction)
print(prission_test)
```

```python
# recall
# it is the ration of number of true positive to accutal poeitive
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
recaal_train=recall_score(y_train,x_train_prediction)
print('recall_score on train data=',recaal_train)
recall_test= recall_score(y_test,x_test_prediction)
print('recall-score on  teast data=',recall_test)
```

```python
# precision .recall and f1 score function
def presion_recall_f1_score(true_label,pred_label):
    priscion_value=precision_score(true_label,pred_label)
    recall_value=recall_score(true_label,pred_label)
    f1_value=f1_score(true_label,pred_label)

    print('predion',priscion_value)
    print('recall',recall_value)
    print('f1',f1_value)
    
```

```python
# clasifition matrix for training data
print('this is for test data')
print('for test data=', presion_recall_f1_score(y_test,x_test_prediction))
```
