```python
# marging the auther name and news titlel
df.isnull().sum()
```

```python
# merging the authoe name and news title
df['content']=df['title']
print(df['content'])
```

```python
# sprating the data & label
x=df.drop(columns='label',axis=1)
y=df['label']
```

```python
print(x)
print(y)
```

```python
# steeming :
# it the prosees of reducing a word to its root word
port_stem=PorterStemmer()
def stemming(content):
    steem_content=re.sub('[^a-zA-Z]',' ',str(content))
    steem_content=steem_content.lower()
    steem_content=steem_content.split()
    steem_content=[port_stem.stem(word) for word in steem_content if not word in stopwords.words('english')]
    steem_content=' '.join(steem_content)
    return steem_content
```

```python
df['content']=df['content'].apply(stemming)
print(df['content'])
```

```python
# speratting the dataat and labelx
x=df['content'].values
y=df['label'].values
```

```python
# convert text datat into numerical dtaat
vectorizer=TfidfVectorizer()
vectorizer.fit(x)

x=vectorizer.transform(x)
print(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
train_dat_pred=model.predict(x_train)
acc_on_train_data=accuracy_score(train_dat_pred,y_train)
```

```python
print(acc_on_train_data)
```

```python
test_data_ored=model.predict(x_test)
acc_on_test_dat=accuracy_score(y_test,test_data_ored)
```
